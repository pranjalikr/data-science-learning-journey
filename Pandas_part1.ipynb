```python
arr = pd.array([10,20,30,40])
print(type(arr))
```

```python
s = pd.Series([10,20,30],index=['val1','val2','val3'])
print(s)
print(s['val1'])
print(s.iloc[0])
```

```python
s = pd.Series([10,20,30,40,50,60],index=['val1','val2','val3','val4','val5','val6'])
print(s)
print(s[0:4])
print(s['val1':'val4'])

print('using for loop')
for i in s:
    print(i)
```

```python
s = pd.Series([10,20,30,40,50,60],index=['val1','val2','val3','val4','val5','val6'])
print(s)
print(s.get('val5'))  #for printing value of that index
print(s.values)   #all the values
print(s.shape)    # whether the array is 1d or 2d
print(s.index)    #print all the index 
print(s.dtype)

#perform numericla computation
print('\nperform numerical computation')
print(s*2)
```

```python
df = pd.read_csv("globalAirQuality.csv")
df
```

```python
#selecting data

df['country']
```

```python
df[["country","aqi"]]
```

```python
#loc-locaation of any particular row through their index and it is label based
#we can use column name directly
#starting-ending(exclusive)
df.loc[3:6]
```

```python
#iloc-location of particular row through index and it is integer(numeric) based
#we have to use column position to get values
#starting-(ending-1)(inclusive)
df.iloc[2:5]
```

```python
#Filtering of Data

df[(df["aqi"]>100) & (df["temperature"] > 30)]

df[df["aqi"] >100 ][["city","aqi"]]
```
